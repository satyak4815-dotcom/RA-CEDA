```python
import os
import subprocess
from pathlib import Path

print("=== KAGGLE ↔ GITHUB CHECK ===")

# 1. Check whether this notebook directory is a Git repository
result = subprocess.run(
    ["git", "rev-parse", "--show-toplevel"],
    capture_output=True, text=True
)

if result.returncode == 0:
    repo = result.stdout.strip()
    print("✅ Git repository found:", repo)

    def git(*args):
        return subprocess.run(
            ["git", *args],
            cwd=repo,
            capture_output=True,
            text=True
        )

    remote = git("remote", "-v")
    print("\nGit remotes:")
    print(remote.stdout if remote.stdout else "No Git remote configured.")

    branch = git("branch", "--show-current")
    print("Branch:", branch.stdout.strip() or "(detached/unknown)")

    status = git("status", "--short")
    print("\nUncommitted changes:")
    print(status.stdout if status.stdout else "None detected.")

else:
    print("⚠️ This notebook directory is not a Git repository.")
    print("Kaggle's GitHub integration may still be configured in its UI.")

# 2. Check whether GitHub credentials appear to be configured
token_present = any(
    os.environ.get(k)
    for k in ("GITHUB_TOKEN", "GH_TOKEN", "GITHUB_PAT")
)
print("\nGitHub token environment variable detected:", token_present)
print("(False does not necessarily mean GitHub is disconnected.)")

# 3. Explain Kaggle persistence
print("\n=== IMPORTANT ===")
print("Notebook edits are not necessarily committed or pushed to GitHub.")
print("Check Kaggle's GitHub integration and repository for the latest commit.")
print("Files saved in /kaggle/working are temporary unless backed up.")
```
